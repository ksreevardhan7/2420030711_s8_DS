```python
 import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
# Load Iris dataset
iris = load_iris()
# Create DataFrame
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df['species'] = iris.target
# 1. Univariate Visualizations
#Histogram
plt.figure(figsize=(6, 4))
sns.histplot(df['sepal length (cm)'], bins=20, kde=True)
plt.title("Histogram of Sepal Length")
plt.xlabel("Sepal Length (cm)")
plt.ylabel("Frequency")
plt.show()
#Boxplot
plt.figure(figsize=(6, 4))
sns.boxplot(y=df['sepal length (cm)'])
plt.title("Boxplot of Sepal Length")
plt.ylabel("Sepal Length (cm)")
plt.show()
#Pie Chart
species_counts = df['species'].value_counts()
plt.figure(figsize=(6, 6))
plt.pie(species_counts,labels=iris.target_names,autopct='%1.1f%%',startangle=90)
plt.title("Distribution of Iris Species")
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
# Load Iris dataset
iris = load_iris()
# Create DataFrame
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df['species'] = iris.target
# 2. Bivariate Visualizations
# Scatter Plot
plt.figure(figsize=(6, 4))
sns.scatterplot(data=df,x='sepal length (cm)',y='sepal width (cm)',hue='species',palette='Set1')
plt.title("Scatter Plot of Sepal Length vs Sepal Width")
plt.xlabel("Sepal Length (cm)")
plt.ylabel("Sepal Width (cm)")
plt.show()
# Line Chart
plt.figure(figsize=(6, 4))
sns.lineplot(data=df,x=df.index,y='sepal length (cm)')
plt.title("Line Chart of Sepal Length")
plt.xlabel("Observation")
plt.ylabel("Sepal Length (cm)")
plt.show()
# Bar Chart - Average Sepal Length by Species
plt.figure(figsize=(6, 4))
sns.barplot(data=df, x='species', y='sepal length (cm)', palette='Set1')
plt.title("Average Sepal Length by Species")
plt.xticks([0, 1, 2], iris.target_names)
plt.xlabel("Species")
plt.ylabel("Average Sepal Length (cm)")
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
iris = load_iris()
df = pd.DataFrame(data=iris.data, columns=iris.feature_names)
df['species'] = iris.target
df['species_name'] = df['species'].map({i:name for i, name in enumerate(iris.target_names)})
#HeatMap
plt.figure(figsize=(6,4))
sns.heatmap(df.iloc[:, :-1].corr(), annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap of Iris Features")
plt.show()
#Bubble plot
plt.figure(figsize=(6,4))
plt.scatter(df['sepal length (cm)'], df['sepal width (cm)'], s = df['petal length (cm)'] * 20, alpha=0.5, c=df['species'], cmap='viridis')
plt.title("Bubble Plot of Sepal Length vs Sepal Width")
plt.xlabel("Sepal Length (cm)")
plt.ylabel("Sepal Width (cm)")
plt.show()
#Pair plot
sns.pairplot(df.iloc[:, :4])
plt.suptitle("Pair Plot of Iris Features, y = 1.02")
plt.show()
```

```python
# Violin Plot - Sepal Length by Species
plt.figure(figsize=(6, 4))
sns.violinplot(x=df['species'],
    y=df['sepal length (cm)']
)
plt.title("Violin Plot of Sepal Length by Species")
plt.xlabel("Species")
plt.ylabel("Sepal Length (cm)")
plt.xticks([0, 1, 2], iris.target_names)
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
# Load Tips dataset
df = pd.read_csv("tips.csv")
# Display first 5 rows
print(df.head())
# 1. Univariate Visualizations
# Histogram
plt.figure(figsize=(6, 4))
sns.histplot(df['total_bill'], bins=20, kde=True)
plt.title("Histogram of Total Bill")
plt.xlabel("Total Bill ($)")
plt.ylabel("Frequency")
plt.show()
# Boxplot
plt.figure(figsize=(6, 4))
sns.boxplot(y=df['total_bill'])
plt.title("Boxplot of Total Bill")
plt.ylabel("Total Bill ($)")
plt.show()
# Pie Chart
sex_counts = df['sex'].value_counts()
plt.figure(figsize=(6, 6))
plt.pie(
    sex_counts,
    labels=sex_counts.index,
    autopct='%1.1f%%',
    startangle=90
)
plt.title("Distribution of Customers by Sex")
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
# Load Tips dataset
df = pd.read_csv("tips.csv")
# 2. Bivariate Visualizations
# Scatter Plot
plt.figure(figsize=(6, 4))
sns.scatterplot(data=df,x='total_bill',y='tip',hue='sex',palette='Set1')
plt.title("Scatter Plot of Total Bill vs Tip")
plt.xlabel("Total Bill ($)")
plt.ylabel("Tip ($)")
plt.show()
# Line Chart
plt.figure(figsize=(6, 4))
sns.lineplot(data=df,x=df.index,y='total_bill')
plt.title("Line Chart of Total Bill")
plt.xlabel("Observation")
plt.ylabel("Total Bill ($)")
plt.show()
# Bar Chart - Average Tip by Day
plt.figure(figsize=(6, 4))
sns.barplot(data=df,x='day',y='tip',palette='Set1')
plt.title("Average Tip by Day")
plt.xlabel("Day")
plt.ylabel("Average Tip ($)")
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
# Load Tips dataset
df = pd.read_csv("tips.csv")
# HeatMap
plt.figure(figsize=(6, 4))
sns.heatmap(df.select_dtypes(include='number').corr(),annot=True,cmap="coolwarm")
plt.title("Correlation Heatmap of Tips Dataset")
plt.show()
# Bubble Plot
plt.figure(figsize=(6, 4))
plt.scatter(df['total_bill'],df['tip'],s=df['size'] * 50,alpha=0.5,c=df['size'],cmap='viridis')
plt.title("Bubble Plot of Total Bill vs Tip")
plt.xlabel("Total Bill ($)")
plt.ylabel("Tip ($)")
plt.show()
# Pair Plot
sns.pairplot(df.select_dtypes(include='number'))
plt.suptitle("Pair Plot of Tips Dataset", y=1.02)
plt.show()
```

```python
# Violin Plot - Tip by Day
plt.figure(figsize=(6, 4))
sns.violinplot(data=df,x='day',y='tip')
plt.title("Violin Plot of Tip by Day")
plt.xlabel("Day")
plt.ylabel("Tip ($)")
plt.show()
```
