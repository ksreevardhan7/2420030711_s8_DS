```python
import numpy as np
from scipy.spatial import distance
pointA = np.array([2, 4, 6])
pointB = np.array([5, 1, 9])
# Euclidean distance
euclidean_dist = distance.euclidean(pointA, pointB)
print("Euclidean Distance :", euclidean_dist)

# similarity (inverse of distance)
similarity_euclidean = 1 / (1 + euclidean_dist)
print("Euclidean similarity :", similarity_euclidean)
```

```python
pointA = np.array([9, 7, 5])
pointB = np.array([3, 2, 8])
manhattan_dist = distance.cityblock(pointA, pointB)
print("Manhattan Distance :", manhattan_dist)
similarity_manhattan = 1 / (1 + manhattan_dist)
print("Manhattan similarity :", similarity_manhattan)
```

```python
import numpy as np
from scipy.spatial import distance
pointA = np.array([9, 7, 5])
pointB = np.array([3, 2, 8])
# minkowski distance with p=3
minkowski_dist_p3 = distance.minkowski(pointA, pointB, p=3)
print("Minkowski Distance (P=3) :", minkowski_dist_p3)
# similarity (inverse of distance)
similarity_minkowski = 1 / (1 + minkowski_dist_p3)
print("Minkowski similarity :", similarity_minkowski)
```

```python
import pandas as pd
from scipy.stats import spearmanr

df = pd.DataFrame({"x": [10, 20, 30, 40, 50], "y": [12, 24, 33, 45, 60]})

corr_value, p_value = spearmanr(df["x"], df["y"])
print(f"Spearman correlation coefficient: {corr_value}")
print(f"P_value: {p_value}")
```

```python
import pandas as pd
from scipy.stats import spearmanr
df = pd.DataFrame(
    {
        "DS": [10, 20, 30, 40, 50],
        "CD": [12, 24, 33, 45, 60],
        "ASE": [15, 25, 35, 42, 55],
    })
corr_matrix = df.corr(method="spearman")
print("Spearman correlation matrix :\n")
print(corr_matrix)
# Correlation b/w DS & CD
corr_DS_CD, p_DS_CD = spearmanr(df["DS"], df["CD"])
print("\nDS vs CD :")
print("Correlation :", corr_DS_CD)
print("P-Value :", p_DS_CD)
# Correlation b/w DS & ASE
corr_DS_ASE, p_DS_ASE = spearmanr(df["DS"], df["ASE"])
print("\nDS vs ASE :")
print("Correlation :", corr_DS_ASE)
print("P-Value :", p_DS_ASE)
# Correlation b/w CD & ASE
corr_CD_ASE, p_CD_ASE = spearmanr(df["CD"], df["ASE"])
print("\nCD vs ASE :")
print("Correlation :", corr_CD_ASE)
print("P-Value :", p_CD_ASE)
```

```python
def hamming_distance(str1, str2):
    if len(str1) != len(str2):
        raise ValueError("Strings must be of equal length")
    return sum(ch1 != ch2 for ch1, ch2 in zip(str1, str2))
s1 = "karolin"
s2 = "kathrin"
dist = hamming_distance(s1, s2)
print(f"Hamming distance between '{s1}' and '{s2}': {dist}")
```

```python
def jaccard_index(str1, str2):
    set1 = set(str1.split())
    set2 = set(str2.split())
    intersection = set1.intersection(set2)
    union = set1.union(set2)
    return len(intersection) / len(union)
s1 = "data science is fun"
s2 = "science makes data useful"
print("Jaccard index :", jaccard_index(s1, s2))
```
