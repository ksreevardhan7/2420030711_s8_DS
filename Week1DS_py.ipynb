```python
#Create a 2D NumPy Array
import numpy as np
n = np.array([[1, 2, 3], [4, 5, 6]])
print(n)
```

```python
#Find the Shape
import numpy as np
n = np.array([[1, 2, 3], [4, 5, 6]])
print("n.shape:", n.shape)
```

```python
#Find the Number of Dimensions
import numpy as np
n = np.array([[1, 2, 3], [4, 5, 6]])
print("n.ndim:", n.ndim)
```

```python
#Find the Type of Array
import numpy as np
n = np.array([[1, 2, 3], [4, 5, 6]])
print("type(n):", type(n))
```

```python
#Reshape a 1D Array into a Row Vector
import numpy as np
m = np.array([1, 2, 3])
m = m.reshape(1, 3)
print(m)
```

```python
#Concatenate Two Arrays
import numpy as np
n = np.array([[1, 2, 3], [4, 5, 6]])
m = np.array([[1, 2, 3]])
x = np.concatenate((n, m))
print(x)
```

```python
#Split an Array
import numpy as np
n = np.array([[1, 2, 3], [4, 5, 6]])
sp = np.split(n, 2)
print(sp)
```

```python
#Find the Index of an Element
import numpy as np
n = np.array([[1, 2, 3], [4, 5, 6]])
s = np.where(n == 3)
print(s)
```

```python
#Sort an Array
import numpy as np
n = np.array([[1, 2, 3], [4, 5, 6]])
p = np.sort(n)
print(p)
```

```python
#Create an Array of Zeros
import numpy as np
n = np.zeros((2, 3))
print(n)
```

```python
#Loop Through a 2D Array
import numpy as np
n = np.array([[1, 2, 3], [4, 5, 6]])
for i in n:
    print(i)
```

```python
#Loop Through a 1D Array
import numpy as np
m = np.array([1, 2, 3])
for i in m:
    print(i)
```

```python
#Arithmetic Operations on NumPy Arrays
import numpy as np
a = np.array([10, 20, 30])
b = np.array([2, 5, 10])
print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Power:", a ** b)
print("Modulus:", a % b)
```

```python
#Reshape
import numpy as np
a = np.array([1, 2, 3, 4])
b = a.reshape(2, 2)
print(b)
```

```python
#Slicing a 1D Array
import numpy as np
a = np.array([10, 20, 30, 40, 50])
print(a[1:4])
```

```python
#Slicing a 2D Array
import numpy as np
a = np.array([[1, 2], [3, 4]])
print(a[0])
```
