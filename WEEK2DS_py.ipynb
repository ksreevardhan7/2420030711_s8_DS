```python
import pandas as pd
import numpy as np
# Sample dataset
df = pd.DataFrame({'Age': [25, 30, np.nan, 40, 35],'Department': ['HR', 'Finance', np.nan, 'IT', np.nan]
})
print(df)
# Mean for numeric column
df['Age'] = df['Age'].fillna(df['Age'].mean())
# Mode for categorical column
df['Department'] = df['Department'].fillna(df['Department'].mode()[0])
print(df)
```

```python
import pandas as pd
import numpy as np
# Sample dataset
df = pd.DataFrame({'Age': [25, 30, np.nan, 40, 35],'Department': ['HR', 'Finance', np.nan, 'IT', np.nan]})
# Display original dataset
print("Original DataSet")
print(df)
#Drop rows with missing values
df_drop_rows = df.dropna()
print("After dropping rows:\n",df_drop_rows)
```

```python
#Removing Duplicates
import pandas as pd
# Sample dataset with duplicates
df = pd.DataFrame({'ID': [1, 2, 2, 3, 4, 4],'Name': ['Alice', 'Bob', 'Bob', 'Charlie', 'David', 'David'],'Age': [25, 30, 30, 35, 40, 40]})
print("Original data:")
print(df)
# Remove exact duplicates
df_exact = df.drop_duplicates()
print("\nAfter Exact Match Removal:")
print(df_exact)
```

```python
# Subset-based removal
# Removing duplicates based on selected key columns
import pandas as pd
# Sample dataset with duplicates
df = pd.DataFrame({'ID': [1, 2, 2, 3, 4, 4],'Name': ['Alice', 'Bob', 'Bob', 'Charlie', 'David', 'David'],'Age': [25, 30, 30, 35, 40, 40]})
print("Original data:")
print(df)
# Remove duplicates based on ID
df_subset_id = df.drop_duplicates(subset=['ID'])
print("\nAfter subset based removal (ID):\n", df_subset_id)
# Remove duplicates based on Name
df_subset_name = df.drop_duplicates(subset=['Name'])
print("\nAfter subset based removal (Name):\n", df_subset_name)
```

```python
# Correcting Inconsistent Formats
import pandas as pd
# Sample dataset with inconsistent data formats
df = pd.DataFrame({'Date':['2025-01-05','05/01/2025','Jan 5,2025','2025.01,05']})
print("Original Data:\n",df)
# Convert all dates into standard ISO format(YYYY-MM-DD)
df['Date'] = pd.to_datetime(df['Date'], errors='coerce').dt.strftime('%Y-%m-%d')
print(df)
```

```python
# Case normalization(lowercase/uppercase)
# Sample dataset with inconsistent cases
df = pd.DataFrame({'Name': ['alice', 'BOB', 'charlie', 'DAVID']})
# Convert names to lowercase
df['Name_lower'] = df['Name'].str.lower()
# Convert names to uppercase
df['Name_upper'] = df['Name'].str.upper()
print(df)
```
