```python
import pandas as pd
from sklearn.datasets import load_iris
from statistics import mode
#load iris dataset
iris = load_iris()
df = pd.DataFrame(iris.data, columns=iris.feature_names)
print("First 5 rows of dataset:")
print(df.head())
# Mean (average)
mean_value = df.mean()
print("\nMean values:")
print(mean_value)
# Median
median_value = df.median()
print("\nMedian values:")
print(median_value)
# Mode
mode_value = df.mode().iloc[0]#represents the multiple repeated values
print("\nMode values:")
print(mode_value)
```

```python
import pandas as pd
from sklearn.datasets import load_iris
from statistics import mode
#load iris dataset
iris = load_iris()
df = pd.DataFrame(iris.data, columns=iris.feature_names)
print("First 5 rows of dataset:")
print(df.head())
#Range(mix-min)
range_value = df.max() - df.min()
print("\nRange values:")
print(range_value)
#Variance
variance_value = df.var()
print("\nVariance values:")
print(variance_value)
#standard devation
std_value = df.std()
print("\nStandard deviation values:")
print(std_value)
#interquartile Range(IOR)
Q1=df.quantile(0.25)
Q3=df.quantile(0.75)
IQR=Q3-Q1
print("\nInterquartile Range values:")
print(IQR)
```

```python
import pandas as pd
from sklearn.datasets import load_iris
from statistics import mode
#load iris dataset
iris = load_iris()
df = pd.DataFrame(iris.data, columns=iris.feature_names)
print("First 5 rows of dataset:")
print(df.head())
#skewness(asymmetry of distribution)
skewness_values = df.skew()
print("\nSkewness values:")
print(skewness_values)
#kurtosis(peakness /  tail heaviness)
kurtosis_values = df.kurtosis()
print("\nKurtosis values:")
print(kurtosis_values)
```
